Reading the txt File.

```python
with open("hostel_bois.txt","r",encoding="utf-8") as f :
   
    lines = f.readlines()
    print("_"*50)

    print("Total lines in the txt file:", len(lines))

print("_"*50)
print("To observe raw Data:-")
for i in range (5):
    print(lines[i])
print("_"*50)

```

1) Creating a Chat Parser

```python
messages = []

system_message_count = 0
deleted_message_count = 0
media_count = 0

for line in lines:

    line = line.strip()

    # Skip empty lines
    if line == "":
        continue

    # Split timestamp and remaining content
    seperate = line.split(" - ", 1)

    if len(seperate) < 2:
        continue

    timestamp = seperate[0]
    content = seperate[1]

    # Skip system messages
    if ":" not in content:
        system_message_count += 1
        continue

    # Split sender and actual message
    sender = content.split(":", 1)

    sender_name = sender[0].strip()
    actual_message = sender[1].strip().lower()     # Normalize message ,makes words like DAMN Damn damn same

    # Count media messages
    if actual_message == "<media omitted>":
        media_count += 1

    # Count deleted messages
    elif actual_message == "this message was deleted":
        deleted_message_count += 1

    # Store everything
    messages.append({
        "Timestamp": timestamp,
        "Sender": sender_name,
        "Message": actual_message
    })

print("To Observe Parsed Data:-")

print("First 3 Parsed Messages:\n")


for msg in messages[0:3]:
    print(msg)

print("\n")
print("Successfully parsed 3174 messages from 6 participants over 60 days!!\n")
print("_"*50)
print("Total Messages :", len(messages))
print("System Messages:", system_message_count)
print("Media Messages :", media_count)
print("Deleted Messages:", deleted_message_count)
print("_"*50)
```

4. Activity Heatmap

```python
participants = list(count1.keys()) # to convert dict into list 
row = len(participants)           # number of group members  
column = 24                       #24 hours in a day

import numpy as np

heatmap = np.zeros((row,column),dtype=int)


participants_index = {}
for i , name in enumerate(participants):          #took help of AI to work with enumerate function
    participants_index[name] = i



for msg in messages:

    sender = msg["Sender"]
    timestamp = msg["Timestamp"]
    hour = int(timestamp.split(",")[1].split(":")[0])
    row = participants_index[sender]
    heatmap[row][hour]+=1




print("\n" + "_"*100)
print("\t\t\t\t   ACTIVITY HEATMAP")
print("_"*100)
print("\nShows participant activity across all 24 hours of the day (60-day dataset)")

print("\nTime Reference (24-Hour Clock):-")
print("00 = 12 AM (Midnight) | 06 = 6 AM | 12 = 12 PM (Noon) | 18 = 6 PM | 23 = 11 PM")

print("\nSymbol Reference:-")
print(".  = 0–25% of participant's peak activity")
print("░  = 25–50% of participant's peak activity")
print("▒  = 50–75% of participant's peak activity")
print("█  = 75–100% of participant's peak activity")
print("\n")
print(f"{'Participants':<15}",end="")

for hour in range (24):
    print(f"{hour:02}",end=" ")

print()
print("-"*100)

for i,row in enumerate((heatmap)):
    max_value = max(row)
    print(f"{participants[i]:<15}",end=" ")
    
    for value in row:
        if max_value == 0:
            max_value=1
        ratio = value/max_value
        if ratio< 0.25 :
            symbol = ". "
        elif ratio < 0.50:
            symbol = "░ "
        elif ratio < 0.75:
            symbol = "▒ "
        else:
            symbol = "█ "
        
        print (symbol,end=" ")
    
    print()

print("_"*100)
       
```

6. Determining Response speed and silent streaks.

```python
from datetime import datetime

response_time = {} #intialize empty dict for response time

for person in participants:
    response_time[person] = [] #empty list for participants

for i in range(len(messages)-1):
    current_message = messages[i]
    next_message = messages[i+1]

    if current_message["Sender"] == next_message["Sender"]:     #skip if same person send the texts
        continue
    
    time_format = "%d/%m/%y, %H:%M"

    current_time = datetime.strptime(current_message["Timestamp"], time_format)
    next_time = datetime.strptime(next_message["Timestamp"], time_format)

    
    time_gap = (next_time - current_time).total_seconds()

    response_time[next_message["Sender"]].append(time_gap)

print(response_time)

```

```python
avg_response = {}

from datetime import timedelta

for person in response_time:
    times = response_time[person]
    
    if len(times)>0:
        avg = sum(times)/len(times)
        avg_response[person] =  avg


avg_response_minutes = {}
for person in avg_response:
    minutes = avg_response[person]/60
    avg_response_minutes[person] = minutes


fastest = min(avg_response_minutes, key = avg_response_minutes.get)
slowest = max(avg_response_minutes, key = avg_response_minutes.get)

current_date = {}
for person in participants:
    current_date[person] = set()

for msg in messages:

    person = msg["Sender"]
    dates = msg["Timestamp"].split(",")[0]
    current_date[person].add(dates)


start_date = datetime.strptime(messages[0]["Timestamp"].split(",")[0], "%d/%m/%y")
end_date = datetime.strptime(messages[-1]["Timestamp"].split(",")[0], "%d/%m/%y")

silent_streak = {}
for person in participants :
    current_day = start_date 
    
    current_streak = 0
    longest_streak = 0 

    while current_day<= end_date:
        date2 = current_day.strftime("%d/%m/%y")

        if date2 in current_date[person] :
            current_streak = 0

        else:
            current_streak+=1

        if current_streak> longest_streak:
            longest_streak = current_streak

        current_day += timedelta (days=1)
    silent_streak [person] = longest_streak

sorted_streak = sorted(
    silent_streak.items(),
    key=lambda x: x[1],
    reverse=True
)

print("_"*70)
print("\t\t       RESPONSE SPEED ")
print("_"*70)
print(f"{'Fastet Replier':<15} : {fastest}, Average Time taken : {avg_response_minutes[fastest]}")
print(f"{'Slowest Replier':<10} : {slowest}, Average Time taken : {avg_response_minutes[slowest]}")
print("_"*70)

print()
print("_"*70)
print("\t\t    Longest Silent Streak")
print("_"*70)

for person , streak in sorted_streak:
    print(f"{person :<5} : {streak}")
print("-"*70)
print(f"Conclusion ~ {sorted_streak[0][0]} is the least active member of the group during \n\t     {first_date.split(",")[0]}  and {last_date.split(",")[0]}")
print("_"*70)
```

7. Personality Archtypes Detection 


a) Night Owl 


```python
night_owl = {}

for i,person in enumerate(participants):
    night_messages = (heatmap[i][23] + heatmap[i][0] + heatmap[i][1] + heatmap[i][2] + heatmap[i][3] + heatmap[i][4])
    total_messages = sum(heatmap[i])

    if total_messages> 0:
        percentage2 = (night_messages/total_messages)*100

    else:
        percentage2 =0

    night_owl[person] = percentage2

print(night_owl)

night_owl_god = max(night_owl, key=night_owl.get)
print(night_owl_god,":", night_owl[night_owl_god], "% messages after 11 pm")
```

c) The Spammer of the Group

```python
spam_streak = {}

for person in participants :
    spam_streak[person] = []

current_sender = messages[0]["Sender"]
current_streak = 1

for i in range(1,len(messages)):
    sender = messages[i]["Sender"]

    if sender == current_sender :
        current_streak +=1 
    else :
        spam_streak[current_sender].append(current_streak)

        current_sender = sender 
        current_streak = 1  # this starts a new streak 

spam_streak[current_sender].append(current_streak) # saves latest streak 

longest_streak = {}

for person in participants:

    if len(spam_streak[person]) > 0:
        longest_streak[person] = max(spam_streak[person])
    else:
        longest_streak[person] = 0
sorted_spam = sorted(
    longest_streak.items(),
    key=lambda x: x[1],
    reverse=True
)


average_streak = {}

for person in participants:
    streaks = spam_streak[person]      # find avg streak for each person

    if len(streaks) > 0:
        average_streak[person] = sum(streaks) / len(streaks)   # total streaks divide bt length of streaks
    else:
        average_streak[person] = 0

spammer = max(average_streak , key=average_streak.get)
spam_avg = average_streak[max(average_streak , key=average_streak.get)]

burst_messager1 = sorted_spam[0][0]
burst_messager2 = sorted_spam[1][0]

burst_messages1 = sorted_spam[0][1]
burst_messages2 = sorted_spam[1][1]


print("_"*70)
print("Spammers of the Group ")
print("-"*60)
print(f"1) Consistent Spammer is {max(average_streak , key=average_streak.get) } with spam average of {average_streak[max(average_streak , key=average_streak.get)]:.2f}")
print(f"2) Burst Messager is {sorted_spam[0][0]} with {sorted_spam[0][1]} messages in a single go")
print("_"*70)
print(sorted_spam)


```

d) The storyteller of Group

```python
total_words = {}         

total_msgs = {}          

for person in participants:
    total_words[person] = 0
    total_msgs[person] = 0

for msg in messages:

    person = msg["Sender"]
    message = msg["Message"]

    words = message.split()        # split the msg into words
    total_words[person] += len(words)    
    total_msgs[person] += 1              

# Calculate average
average_words = {}

for person in participants:
    average_words[person] = total_words[person] / total_msgs[person]

print(average_words)

storyteller = max(average_words, key=average_words.get)

print(f"{storyteller} : The Storyteller of the Group ({average_words[storyteller]:.1f} words/message)")
```

e) The Mom of the Group 

```python
care_words = ['okay', 'safe', 'eat', 'sleep', 'take care', 'are you', 'please', 
'reminder', 'drink water', 'don''t''forget'  ]
    

     


care_score = {}

for person in participants:   
    care_score[person] = 0  

for msg in messages:      
    person = msg["Sender"]
    message = msg["Message"].lower()      
    words = message.split()                
    for word in words:
        if word in care_words:          
            care_score[person] += 1

print(care_score)

group_mom = max(care_score, key=care_score.get)

print(f"{group_mom} : The Mom of the Group with {care_score[group_mom]} caring words")
```

f) The Drama Queen of the Group 

```python
drama_score = {}

for person in participants:    
    drama_score[person] = 0

for msg in messages:       
    person = msg["Sender"]
    message = msg["Message"]

    if message.isupper():          # check for all capital letters 
        drama_score[person] += 1

    elif message.count("!") >= 2:     # checks for exclamaition marks
        drama_score[person] += 1

    

print(drama_score)

drama_queen = max(drama_score, key=drama_score.get)

print(f"{drama_queen} : The Drama Queen of the Group with {drama_score[drama_queen]} dramatic messages ")
```

g) The Comedian of the Group (One who sends most funny words) 


```python
comedy_words = [                                                     # to assign food related words list
    "lol","lmao","lmfao" , "rofl"]
     



comedy_score = {}

for person in participants:       
    comedy_score[person] = 0      

for msg in messages:             

    person = msg["Sender"]
    message = msg["Message"].lower()
    words = message.split()             

    for word in words:                 
        if word in comedy_words:            
            comedy_score[person] += 1     
print(comedy_score)
     


comedy = max(comedy_score, key=comedy_score.get)

print(f"{comedy} : The Comedian of the Group with {comedy_score[comedy]} funny words")
```

h) The Question master of the Group (One who asks most number of questions)

```python
total_messages = {}
question_messages = {}

for person in participants:
    total_messages[person] = 0
    question_messages[person] = 0

for msg in messages:

    person = msg["Sender"]
    message = msg["Message"].strip()

    total_messages[person] += 1

    if message.endswith("?"):
        question_messages[person] += 1

question_percentage = {}

for person in participants:

    if total_messages[person] > 0:
        percentage = (question_messages[person] / total_messages[person]) * 100
    else:
        percentage = 0

    question_percentage[person] = percentage

question_master = max(question_percentage, key=question_percentage.get)



print("_" * 65)
print("\t\t THE QUESTION MASTER")
print("_" * 65)

for person in participants:
    print(
        f"{person:<15} "
        f"{question_messages[person]:>3}/{total_messages[person]:<3} "
        f"({question_percentage[person]:.2f}%)"
    )

print("-" * 65)

if question_percentage[question_master] >= 25:
    print(
        f" {question_master} is the Question Master "
        f"with {question_percentage[question_master]:.2f}% question messages."
    )
else:
    print("No participant qualifies as the Question Master (minimum 25%).")

print("_" * 65)
```
